# Grid Analysis Exploratory Data Analysis

This notebook analyses the synthetic CS112 electricity-grid datasets.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

root = Path('../..')
substations = pd.read_csv(root / 'data/processed/substations_clean.csv')
lines = pd.read_csv(root / 'data/processed/lines_clean.csv')
centrality = pd.read_csv(root / 'data/processed/centrality_results.csv')
```

## Dataset overview


```python
print('Substations:', substations.shape)
print('Lines:', lines.shape)
display(substations.head())
```

## Substations by region


```python
substations['Region'].value_counts().plot(kind='bar', title='Substations by Region')
plt.ylabel('Number of Substations')
plt.tight_layout()
plt.show()
```

## Substation capacity distribution


```python
substations['Capacity (MVA)'].plot(kind='hist', bins=12, title='Substation Capacity Distribution')
plt.xlabel('Capacity (MVA)')
plt.tight_layout()
plt.show()
```

## Transmission-line status


```python
lines['Status'].value_counts().plot(kind='bar', title='Transmission Line Status')
plt.ylabel('Number of Lines')
plt.tight_layout()
plt.show()
```

## Infrastructure age


```python
substations['Asset Age'] = 2026 - substations['Commissioning Year']
substations.groupby('Region')['Asset Age'].mean().sort_values(ascending=False)
```

## Most structurally important substations


```python
centrality[['Substation Name', 'Region', 'Degree', 'Betweenness Centrality']].head(10)
```

## Interpretation note

Centrality measures describe structural importance in the synthetic graph. They do not directly measure real electrical load, voltage stability or power flow.
